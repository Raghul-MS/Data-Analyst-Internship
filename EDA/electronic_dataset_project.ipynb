```python
product_sales=df.groupby("product_name")["number_of_ordered_items"].sum()
product_sales
```

### Observation

- Calculated the total number of ordered items for each product by grouping the dataset by `product_name`.
- This gives me a clear view of which products are selling the most and helps identify top-performing items.

```python
a=product_sales.plot(kind="barh",color="pink",figsize=(10, 6))
plt.xlabel("Total product sales")
plt.ylabel("Product")
plt.title("Product oredered in total (HISTOGRAM BAR CHART)")
for index,value in enumerate(product_sales):
    plt.text(value,index,str(value),ha="left",va="center",fontsize=10,fontweight="bold")
plt.tight_layout()
plt.show()
```

### Observation

- Created a horizontal bar chart to visualize total product sales, making it easier to compare item performance.
- The chart clearly highlights which products have the highest and lowest order volumes.
- Annotating each bar with its value improves readability and helps communicate exact sales figures.
- This visualization supports quick decision-making by showing product demand at a glance.

```python
top_5=df.groupby("product_name")["number_of_ordered_items"].sum().sort_values(ascending=False).head(5)
top_5
```

### Observation

- Identified the top 5 products based on total ordered quantity.
- These products represent the highest demand and are likely key contributors to overall sales.
- This insight helps prioritize inventory, marketing focus, and further analysis on customer preferences.

```python
b=top_5.plot(kind="bar",color="red")
plt.xlabel("PRODUCT NAME")
plt.ylabel("PRODUCT ORDERED")
plt.title("Bar Chart")
for index,value in enumerate(top_5):
    plt.text(index,value,str(value),ha="center",va="bottom",fontsize=8,fontweight="bold")
plt.show()
```

### Observation

- Visualized the top 5 products using a vertical bar chart to highlight their total order volumes.
- The chart clearly distinguishes high-performing items, making it easy to compare product popularity.
- Annotating each bar with its exact value improves clarity and supports quick interpretation.
- This visualization helps in identifying which products drive the most sales and may deserve strategic focus.

```python
month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

monthly_counts = df.groupby("Month")["product_name"].count().reindex(month_order, fill_value=0)
monthly_counts
```

### Observation

- Grouped the data by `Month` and counted the number of product entries per month.
- Reindexing with a full month list ensures chronological order, even if some months have no data.
- This helps identify seasonal trends, peak activity periods, and months with low engagement.
- The `fill_value=0` ensures that missing months are represented with zero, maintaining visual and analytical consistency.

```python
monthly_counts.plot(marker="*",linestyle="--",figsize=(10,6),color="Blue")
plt.xlabel("MONTH")
plt.ylabel("ITEMS ORDER IN MONTHWISE")
plt.title("LINE CHART")
plt.show()
```

### Observation

- Plotted a line chart to visualize monthly product order counts, using asterisk markers and dashed lines for emphasis.
- The chart reveals fluctuations in order volume across months, helping identify seasonal peaks and dips.
- Months with zero orders are clearly visible, indicating inactive periods or missing data.
- This visualization supports time-based analysis and can guide decisions on inventory planning, promotions, or staffing.

```python
df.groupby("product_name")["order_status"].value_counts()
```

### Observation

- Grouped the dataset by `product_name` and counted the frequency of each `order_status` within each product.
- This breakdown reveals how each product is performing in terms of fulfillment—whether orders are completed, pending, cancelled, or returned.

```python
order_details=df.groupby("order_status")["product_name"].count()
order_details
```

### Observation

- Counted the number of product entries for each `order_status`, giving a clear overview of how orders are distributed across different statuses.
- This helps assess operational efficiency—whether most orders are delivered, In process, cancelled, or on hold.
- A high count of completed orders suggests smooth fulfillment, while elevated cancelled or returned counts may indicate issues in product quality, delivery, or customer satisfaction.
- This summary supports strategic decisions in logistics, customer service, and product management.

```python
labels=["Canceled","Delivered","In Process","On Hold"]
plt.pie(order_details,labels=labels,autopct="%1.1f%%",startangle=90)
plt.title("Order Status in Pie Chart(in percentage)")
plt.show()
```

### Observation

- Created a pie chart to visualize the distribution of order statuses as percentages.
- The chart provides a quick snapshot of how many orders fall into each category: Canceled, Delivered, In Process, and On Hold.
- This helps assess fulfillment efficiency and identify potential issues in the order pipeline.
- A high percentage of delivered orders indicates strong operational performance, while elevated canceled or on-hold rates may require attention.
