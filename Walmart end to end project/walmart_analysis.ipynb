```python
import pandas as pd
features=pd.read_csv("Downloads/Walmart/features.csv")
stores=pd.read_csv("Downloads/Walmart/stores.csv")
test=pd.read_csv("Downloads/Walmart/test.csv")
train=pd.read_csv("Downloads/Walmart/train.csv")
```

```python
#just seen how the data structure is
print(features.head())
print(stores.head())
print(train.head())
print(test.head())
```

```python
#converted date to date time
for df in [features, train, test]:
    df['Date'] = pd.to_datetime(df['Date'])
```

```python
print(features.shape)
print(stores.shape)
print(train.shape)
print(test.shape)
```

```python
print(features.info())
print(stores.info())
print(train.info())
print(test.info())
```

```python
print(features.isnull().sum())
print(stores.isnull().sum())
print(train.isnull().sum())
print(test.isnull().sum())
```

```python
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']
features[markdown_cols] = features[markdown_cols].fillna(0)
```

```python
features['CPI'].fillna(features['CPI'].median(), inplace=True)
features['Unemployment'].fillna(features['Unemployment'].median(), inplace=True)
```

```python
features['IsHoliday'] = features['IsHoliday'].astype(int)
train['IsHoliday'] = train['IsHoliday'].astype(int)
test['IsHoliday'] = test['IsHoliday'].astype(int)
```

```python
train_full = train.merge(stores, on="Store", how="left")
train_full = train_full.merge(features, on=["Store", "Date"], how="left")

test_full = test.merge(stores, on="Store", how="left")
test_full = test_full.merge(features, on=["Store", "Date"], how="left")
```

```python
print(train_full.columns)
print(test_full.columns)
```

```python
print(train_full.head(2))
print(test_full.head(2))
```

```python
print(train_full.isnull().sum())
print(test_full.isnull().sum())
```
